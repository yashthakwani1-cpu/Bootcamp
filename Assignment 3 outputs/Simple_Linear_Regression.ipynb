```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Prepare sample data (X must be a 2D array/matrix, y is a 1D array)
# X = Years of Experience, y = Salary (in thousands)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([45, 50, 60, 65, 70])

# 2. Initialize the model
model = LinearRegression()

# 3. Train the model (This is where the OLS math happens)
model.fit(X, y)

# 4. View the calculated math equation parameters
print(f"Calculated Slope (m): {model.coef_[0]:.2f}")
print(f"Calculated Intercept (b): {model.intercept_:.2f}")
print(f"Calculated Equation: Salary = {model.coef_[0]:.2f} * Experience + {model.intercept_:.2f}\n")

# 5. Predict salary for someone with 3.5 years of experience
new_experience = np.array([[3.5]])
predicted_salary = model.predict(new_experience)
print(f"Predicted Salary for 3.5 years of experience: ${predicted_salary[0]:.2f}k")
```
